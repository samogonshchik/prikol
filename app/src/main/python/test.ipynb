```python
def getAttrs(clazz):
    return {name: attr for name, attr in clazz.__dict__.items()
            if not name.startswith("__") 
            and not callable(attr)
            and not type(attr) is staticmethod}
```

```python
import CWG
from time import time
import json
import string
from os.path import dirname, join


with open("word_definitions.json",'r') as f:
    all_words = json.load(f)
words = {i:{j:{ch:set() for ch in set(string.ascii_lowercase)} for j in range(0,i)} for i in range(1,18)}

for word in all_words.keys():
    for i in range(len(word)):
        words[len(word)][i][word[i]].add(word)

# generating crossword
cw = CWG.CrosswordGenerator(_vocab=words, _dictionary=all_words, _layout=1)
cw.generate()
board = cw.board
print(board.clues)
```

```python
getAttrs(board)
```
